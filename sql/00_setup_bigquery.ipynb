## 1️⃣ Autenticación y conexión al proyecto de BigQuery
Autenticamos la cuenta de Google y conectamos con el proyecto `tfm-mercado-laboral-datos`, comprobando que el dataset de destino existe.

```python
from google.colab import auth
auth.authenticate_user()
```

```python
PROJECT_ID = "tfm-mercado-laboral-datos"

from google.cloud import bigquery
client = bigquery.Client(project=PROJECT_ID)

for dataset in client.list_datasets():
    print(dataset.dataset_id)
```

## 2️⃣ Carga de la tabla `postings` a BigQuery
Montamos Google Drive, leemos el CSV original de ofertas (123.849 filas) y lo escribimos en BigQuery como punto de partida para el resto del proyecto.

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
import pandas as pd

ruta_archivo = "/content/drive/MyDrive/TFM/postings.csv"
df_postings = pd.read_csv(ruta_archivo)

print(df_postings.shape)
df_postings.head()
```

```python
table_id = f"{PROJECT_ID}.mercado_laboral_datos.postings"

job = client.load_table_from_dataframe(df_postings, table_id)
job.result()  # espera a que la carga termine del todo

print("Filas cargadas en BigQuery:", job.output_rows)
```
